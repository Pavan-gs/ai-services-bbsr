LangChain application engineering**

Understanding the Runnable interface**


**How does LangChain make different components work together consistently?**

Think about the components we've already encountered:

- `ChatPromptTemplate` creates messages.
- `ChatGoogleGenerativeAI` accepts messages and generates a response.
- `StrOutputParser` converts an AI message into a string.
- `JsonOutputParser` converts model output into a Python dictionary.

They perform different jobs, but LangChain gives many of these components a common execution interface called `Runnable`.

### **1.1 First understand the problem**

Consider this ordinary Python implementation:

```
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_template(
    "Explain {topic} in simple terms."
)

llm = ChatGoogleGenerativeAI(
    model="YOUR_CONFIGURED_GEMINI_MODEL"
)

parser = StrOutputParser()

# Execute each component manually
messages = prompt.invoke({"topic": "RAG"})

ai_response = llm.invoke(messages)

final_answer = parser.invoke(ai_response)

print(final_answer)
```

Notice what we're doing.

**Step 1: Prompt**

Input: `{"topic": "RAG"}`

Output: `ChatPromptValue`

**Step 2: LLM**

Input: `ChatPromptValue`

Output: `AIMessage`

**Step 3: Parser**

Input: `AIMessage`

Output: `str`

We're manually taking the output of one component and passing it to the next.

This works perfectly well. But imagine an application with 15 components. We would need to manage the execution, input/output connections, error handling, and potentially parallel operations ourselves.

LangChain provides a common interface to simplify this.

### **1.2 What exactly is a Runnable?**

A Runnable is an abstraction that represents something that can be executed using a consistent set of methods.

Many LangChain components implement this interface.

For example:

| Component                | Input                         | Output                        |
| ------------------------ | ----------------------------- | ----------------------------- |
| `ChatPromptTemplate`     | Dictionary                    | Chat prompt value             |
| `ChatGoogleGenerativeAI` | Messages or prompt value      | AIMessage                     |
| `StrOutputParser`        | AIMessage                     | String                        |
| `JsonOutputParser`       | AIMessage                     | Dictionary                    |
| `RunnableLambda`         | Whatever its function accepts | Whatever its function returns |

The key idea:

**Different components can expose the same execution methods even though their internal responsibilities and data types are different.**

### **1.3 The common execution methods**

Let's understand the methods individually.

**A.&#x20;**`invoke()`**&#x20;— execute once**

```
result = prompt.invoke({    "topic": "Retrieval Augmented Generation"})print(result)
```

It executes the prompt component once and returns the resulting prompt value.

We can also invoke the model:

```
response = llm.invoke(result)print(response.content)
```

And invoke the parser:

```
answer = parser.invoke(response)print(answer)
```

The important distinction: `invoke()` is not exclusive to the LLM. It can be used with prompts, models, parsers, and other Runnable components.

**B.&#x20;**`batch()`**&#x20;— execute multiple inputs**

Suppose we want to generate explanations for three concepts.

```
topics = [    {"topic": "Embeddings"},    {"topic": "Vector databases"},    {"topic": "Semantic search"}]results = prompt.batch(topics)for result in results:    print(result)
```

Here, `batch()` executes the same Runnable against multiple inputs.

Notice that we are invoking the prompt, not the complete LLM pipeline. It will generate three prompt values, not three AI answers.

Later, when we compose a complete pipeline, we can batch the entire pipeline.

**C.&#x20;**`stream()`**&#x20;— receive output incrementally**

```
for chunk in llm.stream(    "Explain RAG briefly."):    print(chunk.content, end="", flush=True)
```

Instead of waiting for the complete response, we receive chunks as the model generates them.

A subtle but important point: streaming behaviour depends on the component. A prompt template generally produces its result immediately; an LLM can stream generated tokens or chunks.

**D.&#x20;**`ainvoke()`**&#x20;— asynchronous execution**

```
import asyncioasync def main():    result = await llm.ainvoke(        "Explain vector databases."    )    print(result.content)asyncio.run(main())
```

This is useful in asynchronous applications, especially when integrating LangChain with FastAPI.

It allows the application to await an operation without blocking the event loop in the same way a synchronous call would.

### **1.4 One important distinction: Runnable does not mean identical input/output**

Suppose we have:

```
prompt.invoke({"topic": "RAG"})
```

But:

```
parser.invoke(ai_message)
```

Both use `invoke()`, but they accept different input types.

The common interface standardizes execution methods, not the business meaning or data types of every component.

This distinction becomes especially important when we connect components using LCEL. The output of one component must be compatible with the input expected by the next component.

For example:

```
Prompt output → Model input → Compatible
Model output  → Parser input → Compatible
```

If the types don't match, we need an appropriate transformation.

## **1.5 Reference table — Runnable interface**

| Method / Component | How we use it                            | What it does                          |
| ------------------ | ---------------------------------------- | ------------------------------------- |
| `Runnable`         | Implemented by many LangChain components | Provides a common execution interface |
| `invoke()`         | `prompt.invoke(input)`                   | Executes once                         |
| `batch()`          | `prompt.batch(inputs)`                   | Executes against multiple inputs      |
| `stream()`         | `llm.stream(input)`                      | Produces output incrementally         |
| `ainvoke()`        | `await llm.ainvoke(input)`               | Asynchronous execution                |
| `abatch()`         | `await prompt.abatch(inputs)`            | Asynchronous batch execution          |
| `astream()`        | `async for chunk in llm.astream(input)`  | Asynchronous streaming                |

**What we have established:** A prompt, model, and parser can all be executed through a common interface. But we are still manually connecting them.

When should we use LCEL?**

Consider these application requirements:

| Requirement                                       | Suitable approach            |
| ------------------------------------------------- | ---------------------------- |
| Prompt → LLM → parser                             | LCEL                         |
| Query → classify → format result                  | LCEL                         |
| Retrieve context → generate answer                | LCEL                         |
| Generate summary and classification independently | LCEL with `RunnableParallel` |
| Multiple conditional paths with complex state     | Consider LangGraph           |
| Retrieve → grade → rewrite → retrieve again       | LangGraph is a better fit    |
| Pause for human approval and resume later         | LangGraph                    |

## Reference table — RunnableSequence and LCEL**

| Method / Component | How we use it                     | What it does                                          |
| ------------------ | --------------------------------- | ----------------------------------------------------- |
| LCEL               | `prompt \\| llm \\| parser`       | Composes Runnables into a pipeline                    |
| `RunnableSequence` | Created through `\\|` composition | Executes Runnables sequentially                       |
| `chain.invoke()`   | `chain.invoke(input_data)`        | Executes the complete pipeline                        |
| `chain.batch()`    | `chain.batch(inputs)`             | Executes the pipeline for multiple inputs             |
| `chain.stream()`   | `chain.stream(input_data)`        | Streams output through the composed pipeline          |
| Nested chains      | `chain_a \\| chain_b`             | Combines reusable pipelines                           |
| Callable coercion  | `chain \\| lambda x: ...`         | Converts supported callables into Runnable components |

**The key takeaway:** LCEL allows us to define how components connect once, and then execute the entire sequence using a common Runnable interface.

Understanding fan-out and fan-in

Imagine a customer sends this query:
I met with an accident yesterday. My car is badly damaged. How do I proceed with my insurance claim?

We want three independent things from this one query:
1. Summarize the query.
2. Classify the query.
3. Recommend next steps.
Fan-out: one input, multiple operations
Think of a single input being distributed to three independent processing branches.

````python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()

# Summary
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize this customer query in two sentences:\n{query}"
)

summary_chain = summary_prompt | llm | parser


# Classification
classification_prompt = ChatPromptTemplate.from_template(
    """
    Classify the insurance query into one category:
    Claim, Policy, Premium, Renewal, or Other.

    Return only the category.

    Query: {query}
    """
)

classification_chain = classification_prompt | llm | parser


# Recommended next steps
steps_prompt = ChatPromptTemplate.from_template(
    """
    Suggest three practical next steps for the customer.

    Do not invent policy-specific requirements.

    Customer query: {query}
    """
)

steps_chain = steps_prompt | llm | parser
````

````python
from langchain_core.runnables import RunnableParallel

analysis_chain = RunnableParallel(
    summary=summary_chain,
    classification=classification_chain,
    next_steps=steps_chain
)
````

````python
customer_input = {
    "query": """
    I met with an accident yesterday.
    My car is badly damaged.
    How do I proceed with my insurance claim?
    """
}

analysis_result = analysis_chain.invoke(customer_input)

print(analysis_result)
````

````python
from langchain_core.runnables import (
    RunnableLambda,
    RunnablePassthrough,
    RunnableParallel
)

def clean_and_validate(data):

    query = data.get("query", "").strip()

    if not query:
        raise ValueError("Query cannot be empty")

    return {
        "query": query
    }


cleaning_runnable = RunnableLambda(clean_and_validate)

analysis_chain = RunnableParallel(
    summary=summary_chain,
    classification=classification_chain,
    next_steps=steps_chain
)

complete_analysis_chain = (
    cleaning_runnable
    | RunnablePassthrough.assign(
        analysis=analysis_chain
    )
)
````

````python
result = complete_analysis_chain.invoke({
    "query": "   I met with an accident yesterday. How do I claim insurance?   "
})
````
