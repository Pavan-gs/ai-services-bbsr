```python
import sqlite3
```

```python
conn = sqlite3.connect('test1.db')
print("The database is opened successfully")
```

```python
conn.execute('''CREATE TABLE EMPLOYEE
             (EID INT PRIMARY KEY NOT NULL,
             NAME             TEXT NOT NULL,
             AGE              INT  NOT NULL,
             ADDRESS          CHAR(80),
             SALARY           REAL);''')
print("Table Employee created successfully")
```

```python
conn.execute("INSERT INTO EMPLOYEE (EID,NAME,AGE,ADDRESS,SALARY)\
             VALUES(1,'Sean',28,'#84 flo',4500)")
conn.execute("INSERT INTO EMPLOYEE (EID,NAME,AGE,ADDRESS,SALARY)\
             VALUES(2,'Jean',24,'#24 txs',2500)")       
conn.execute("INSERT INTO EMPLOYEE (EID,NAME,AGE,ADDRESS,SALARY)\
             VALUES(3,'Kiran',32,'#24 Mum',8500)")        
conn.commit()
print("Records inserted successsfully")
```

```python
# Query the table
conn=sqlite3.connect('test1.db')
sql = "SELECT * from EMPLOYEE;"
cus = conn.cursor()
cus.execute(sql)
emp = cus.fetchall()
for i in emp:
    print(i)
```
