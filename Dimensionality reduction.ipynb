```python
cars1 = cars.iloc[[0,4,8],1:5:3]
```

```python
cars1.reset_index(drop=True,inplace=True)
```

```python
# Make the data mean centric
```

```python
centered_data = cars1-np.mean(cars1.T, axis=1)
```

```python
np.mean(cars1.T, axis=1)
```

```python
cars1
```

```python
centered_data
```

```python
# Standardization --> Center & scale --> (xi-mean(x))/std(x) 
```

```python
# Find the co-variance matrix for the data
```

```python
cov_mat=np.cov(centered_data.T)
```

```python
centered_data
```

```python
cov_mat
```

```python
# Find the corresponding eigen values and vectors
```

```python
from scipy.linalg import eig
```

```python
Eval,Evec = eig(cov_mat)
```

```python
Eval
```

```python
Evec
```

```python
df = pd.read_csv("E:/data/Wine.csv")
```

```python
df.head()
```

```python
df['Customer_Segment'].value_counts()
```

```python
df.isna().sum()
```

```python
df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
round(df.describe(),2).T
```

```python
x = df.iloc[:,:-1]
```

```python
y = df['Customer_Segment']
```

```python
# standardise the data
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
x_std = sc.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_std,y,test_size=0.10, random_state=2)
```

```python
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components=2)
```

```python
xtrain_pca = pca.fit_transform(x_train)
```

```python
xtest_pca = pca.transform(x_test)
```

```python
pca.explained_variance_ratio_
```

```python
pca.components_
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg = LogisticRegression(random_state=2)
```

```python
lg.fit(xtrain_pca,y_train)
```

```python
y_pred_pca = lg.predict(xtest_pca)
```

```python
y_pred_pca
```

```python
pd.crosstab(y_test,y_pred_pca)
```

```python
17/18
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test,y_pred_pca))
```

```python
pd.crosstab(y_train,lg.predict(xtrain_pca))
```

```python
# Visualising the Training set results
from matplotlib.colors import ListedColormap
X_set, y_set = xtrain_pca, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, lg.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green', 'blue')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green', 'blue'))(i), label = j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```

```python
# Visualising the Test set results
from matplotlib.colors import ListedColormap
X_set, y_set = xtest_pca, y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, lg.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green', 'blue')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green', 'blue'))(i), label = j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
```

```python
x = df.iloc[:,[2,3]]
```

```python
y = df['Purchased']
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
sc = StandardScaler()
```

```python
scaledx = sc.fit_transform(x)
```

```python
from sklearn.model_selection import train_test_split
```

```python
xtrain,x_test,y_train,y_test = train_test_split(scaledx,y, test_size=0.10, random_state=2)
```

```python
from sklearn.decomposition import PCA, KernelPCA
```

```python
kpca = KernelPCA(n_components=2, random_state=2,kernel='rbf')
```

```python
xtrain_kpca = kpca.fit_transform(xtrain)
```

```python
xtest_kpca = kpca.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lg = LogisticRegression(random_state=2)
```

```python
lgmodel = lg.fit(xtrain_kpca,y_train)
```

```python
ypred = lgmodel.predict(xtest_kpca)
```

```python
pd.crosstab(ypred,y_test)
```

```python
32/40
```

```python
10/17
```

```python
# Visualising the Training set results
from matplotlib.colors import ListedColormap
X_set, y_set = xtrain_kpca, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, lgmodel.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```

```python
# Visualising the Test set results
from matplotlib.colors import ListedColormap
X_set, y_set = xtest_kpca, y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, lgmodel.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green'))(i), label = j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('Age')
plt.ylabel('Estimated Salary')
plt.legend()
plt.show()
```
