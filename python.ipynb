```python
class emp:
    '''This class provides rules to create employee records'''
    org = 'LTM'
    Loc = 'Kalinga'
    City = 'BBSR'
    count = 0

    def __init__(self,id,name,salary):
        self.id = id
        self.name = name
        self.salary = salary
        emp.count+=1
    
    def greet(self):
        print("Welcome to the {} campus at {}".format(emp.org,emp.Loc))

    def __repr__(self):
        return '{} --> {}'.format(self.__class__.__name__,emp.count)

    def __len__(self):
        return emp.count
```

```python
from abc import ABC, abstractmethod
```

```python
class car(ABC):
    @abstractmethod
    def mileage(self):
        pass
    
    @abstractmethod
    def manufc(self):
        pass
    
class sports(car):
    def mileage(self):
        print("Mileage is about 10 kms")
        
    def manufc(self):
        print("The name of the manufacturer is Honda")
        
class economy(car):
    def mileage(self):
        print("Mileage is about 18 kms")
        
    def manufc(self):
        print("The name of the manufacturer is Honda")
```

```python
e=economy()
```

```python
class book:
    def __init__(self,title,author,pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    def __str__(self):
        return "The {} is written by {} and has {} pages" .format(self.title,self.author,self.pages)
    
    def __len__(self):
        return self.pages
    
    def __del__(self):
        print("The book is deleted")
```
