```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

try:
    response = requests.get(url, timeout=10)

    response.raise_for_status()

    data = response.json()

    print("Title:", data["title"])
    print("Body:", data["body"])

except requests.exceptions.Timeout:
    print("The request timed out.")

except requests.exceptions.HTTPError as exc:
    print("HTTP error:", exc)

except requests.exceptions.RequestException as exc:
    print("Request failed:", exc)

except ValueError:
    print("The response could not be parsed as JSON.")
```

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

headers = {
    "Accept": "application/json"
}

response = requests.get(
    url,
    headers=headers,
    timeout=10
)

response.raise_for_status()

print(response.json())
```

```python

import requests


def fetch_posts(user_id):
    url = "https://jsonplaceholder.typicode.com/posts"

    params = {
        "userId": user_id
    }

    headers = {
        "Accept": "application/json"
    }

    try:
        response = requests.get(
            url,
            params=params,
            headers=headers,
            timeout=10
        )

        response.raise_for_status()

        return response.json()

    except requests.exceptions.Timeout:
        print("The API request timed out.")

    except requests.exceptions.HTTPError as exc:
        print("HTTP error:", exc)

    except requests.exceptions.RequestException as exc:
        print("Request failed:", exc)

    except ValueError:
        print("The API response was not valid JSON.")

    return None


posts = fetch_posts(1)

if posts is not None:
    print("Posts retrieved:", len(posts))

    for post in posts:
        print(post.get("title", "Untitled"))
```

```python

import streamlit as st
import pymupdf

st.title("PDF Text Extractor")

uploaded_file = st.file_uploader(
    "Upload a PDF file",
    type=["pdf"]
)

if uploaded_file is not None:
    try:
        doc = pymupdf.open(
            stream=uploaded_file.getvalue(),
            filetype="pdf"
        )

        extracted_pages = []

        for page_number, page in enumerate(doc, start=1):
            text = page.get_text()
            extracted_pages.append(
                f"Page {page_number}\n{text}"
            )

        doc.close()

        extracted_text = "\n\n".join(extracted_pages)

        st.subheader("Extracted Text")
        st.text_area(
            "PDF content",
            value=extracted_text,
            height=400
        )

    except Exception as exc:
        st.error(f"Could not process the PDF: {exc}")
```

```python

import requests
import streamlit as st

st.title("API Response Viewer")

user_id = st.number_input(
    "Enter a user ID",
    min_value=1,
    max_value=10,
    value=1
)

if st.button("Fetch Posts"):
    url = "https://jsonplaceholder.typicode.com/posts"

    try:
        response = requests.get(
            url,
            params={"userId": int(user_id)},
            timeout=10
        )

        response.raise_for_status()
        posts = response.json()

        st.success(f"Retrieved {len(posts)} posts")

        for post in posts:
            st.subheader(post.get("title", "Untitled"))
            st.write(post.get("body", ""))

    except requests.exceptions.RequestException as exc:
        st.error(f"API request failed: {exc}")

    except ValueError:
        st.error("The API returned invalid JSON.")
```

FastAPI is a Python web framework used to build APIs, especially HTTP APIs. It lets us define endpoints that receive requests, execute Python code and return responses.

Imagine a document-processing application. Streamlit provides the interface where a user uploads a PDF. FastAPI can provide the backend endpoints that accept the document, extract text and return the result.


Client
Browser, Streamlit app, Postman or another service

|
|

FastAPI application
Receives requests and routes them to Python functions

|
|

Application logic
Validate inputs, process data, interact with services or databases

|
|

HTTP response
Status code, headers and response body

FastAPI helps with endpoint routing, request validation, response serialization and automatic API documentation. It does not automatically provide a database, user authentication system or business logic; those need to be added as appropriate.

```python
from fastapi import FastAPI
app = FastAPI()

@app.get("/")
def home():
    return {"message": "Welcome to FastAPI"}

@app.get("/about")
def about():
    return {
        "framework": "FastAPI",
        "purpose": "Building HTTP APIs"
    }

@app.get("/health")
def health_check():
    return {"status": "healthy"}
```

```python
@app.get("/products/{product_id}")
def get_product(product_id: int):
    return {
        "product_id": product_id,
        "message": "Product retrieved"
    }
```

```python
@app.get("/products")
def get_products(category: str = "all", limit: int = 10):
    return {
        "category": category,
        "limit": limit
    }
```
